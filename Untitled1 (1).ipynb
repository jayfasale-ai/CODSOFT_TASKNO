```python

import os
import zipfile
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("Libraries imported successfully!")


# ============================================================
# 2. ZIP FILE PATH
# ============================================================

zip_path = "/content/archive(2).zip"

print("\nZIP file path:")
print(zip_path)


# ============================================================
# 3. CHECK ZIP FILE
# ============================================================

if not os.path.exists(zip_path):
    raise FileNotFoundError(
        f"The file {zip_path} was not found. "
        "Please upload archive(2).zip to /content/."
    )

print("\nZIP file found successfully!")


# ============================================================
# 4. EXTRACT ZIP FILE
# ============================================================

extract_path = "/content/fraud_dataset"

os.makedirs(
    extract_path,
    exist_ok=True
)

print("\nExtracting ZIP file...")

with zipfile.ZipFile(
    zip_path,
    "r"
) as zip_ref:

    zip_ref.extractall(
        extract_path
    )

print("ZIP file extracted successfully!")


# ============================================================
# 5. FIND CSV FILES
# ============================================================

train_path = None
test_path = None

for root, folders, files in os.walk(
    extract_path
):

    for file in files:

        file_lower = file.lower()

        if file_lower == "fraudtrain.csv":

            train_path = os.path.join(
                root,
                file
            )

        elif file_lower == "fraudtest.csv":

            test_path = os.path.join(
                root,
                file
            )


print("\n============================================")
print("FILES FOUND")
print("============================================")

print("Training file:")
print(train_path)

print("\nTesting file:")
print(test_path)


# ============================================================
# 6. CHECK REQUIRED FILES
# ============================================================

if train_path is None:

    raise FileNotFoundError(
        "fraudTrain.csv was not found inside the ZIP file."
    )

if test_path is None:

    raise FileNotFoundError(
        "fraudTest.csv was not found inside the ZIP file."
    )

print("\nRequired CSV files found successfully!")


# ============================================================
# 7. LOAD TRAINING DATA
# ============================================================

print("\n============================================")
print("LOADING TRAINING DATA")
print("============================================")

train = pd.read_csv(
    train_path
)

print("Training data loaded successfully!")

print("Training data shape:")
print(train.shape)

print("\nFirst 5 rows:")
print(train.head())


# ============================================================
# 8. LOAD TESTING DATA
# ============================================================

print("\n============================================")
print("LOADING TEST DATA")
print("============================================")

test = pd.read_csv(
    test_path
)

print("Testing data loaded successfully!")

print("Testing data shape:")
print(test.shape)

print("\nFirst 5 rows:")
print(test.head())


# ============================================================
# 9. CHECK TARGET COLUMN
# ============================================================

print("\n============================================")
print("TARGET COLUMN")
print("============================================")

if "is_fraud" not in train.columns:

    raise ValueError(
        "The column 'is_fraud' was not found in training data."
    )

if "is_fraud" not in test.columns:

    raise ValueError(
        "The column 'is_fraud' was not found in test data."
    )

print("Target column: is_fraud")

print("\nTarget values:")
print(
    train["is_fraud"].value_counts()
)


# ============================================================
# 10. CHECK FRAUD DISTRIBUTION
# ============================================================

print("\n============================================")
print("FRAUD DISTRIBUTION")
print("============================================")

fraud_counts = train["is_fraud"].value_counts()

print(fraud_counts)

print("\nPercentage:")

print(
    train["is_fraud"].value_counts(
        normalize=True
    ) * 100
)


# ============================================================
# 11. REMOVE UNNECESSARY COLUMNS
# ============================================================

print("\n============================================")
print("REMOVING UNNECESSARY COLUMNS")
print("============================================")

columns_to_remove = [
    "Unnamed: 0",
    "cc_num",
    "first",
    "last",
    "street",
    "trans_num"
]

# Only remove columns that exist

train_columns_to_remove = [
    column
    for column in columns_to_remove
    if column in train.columns
]

test_columns_to_remove = [
    column
    for column in columns_to_remove
    if column in test.columns
]

train = train.drop(
    columns=train_columns_to_remove
)

test = test.drop(
    columns=test_columns_to_remove
)

print("Removed columns:")
print(train_columns_to_remove)


# ============================================================
# 12. PROCESS TRANSACTION DATE AND TIME
# ============================================================

print("\n============================================")
print("PROCESSING DATE AND TIME")
print("============================================")

if "trans_date_trans_time" in train.columns:

    train["trans_date_trans_time"] = pd.to_datetime(
        train["trans_date_trans_time"],
        errors="coerce"
    )

    test["trans_date_trans_time"] = pd.to_datetime(
        test["trans_date_trans_time"],
        errors="coerce"
    )


    # Year

    train["transaction_year"] = (
        train["trans_date_trans_time"].dt.year
    )

    test["transaction_year"] = (
        test["trans_date_trans_time"].dt.year
    )


    # Month

    train["transaction_month"] = (
        train["trans_date_trans_time"].dt.month
    )

    test["transaction_month"] = (
        test["trans_date_trans_time"].dt.month
    )


    # Day

    train["transaction_day"] = (
        train["trans_date_trans_time"].dt.day
    )

    test["transaction_day"] = (
        test["trans_date_trans_time"].dt.day
    )


    # Hour

    train["transaction_hour"] = (
        train["trans_date_trans_time"].dt.hour
    )

    test["transaction_hour"] = (
        test["trans_date_trans_time"].dt.hour
    )


    # Minute

    train["transaction_minute"] = (
        train["trans_date_trans_time"].dt.minute
    )

    test["transaction_minute"] = (
        test["trans_date_trans_time"].dt.minute
    )


    # Day of week

    train["transaction_dayofweek"] = (
        train["trans_date_trans_time"].dt.dayofweek
    )

    test["transaction_dayofweek"] = (
        test["trans_date_trans_time"].dt.dayofweek
    )


    # Remove original column

    train = train.drop(
        columns=["trans_date_trans_time"]
    )

    test = test.drop(
        columns=["trans_date_trans_time"]
    )


print("Date and time features processed!")


# ============================================================
# 13. PROCESS DATE OF BIRTH
# ============================================================

print("\n============================================")
print("PROCESSING AGE")
print("============================================")

if "dob" in train.columns:

    train["dob"] = pd.to_datetime(
        train["dob"],
        errors="coerce"
    )

    test["dob"] = pd.to_datetime(
        test["dob"],
        errors="coerce"
    )


    train["age"] = (
        train["transaction_year"]
        - train["dob"].dt.year
    )

    test["age"] = (
        test["transaction_year"]
        - test["dob"].dt.year
    )


    train = train.drop(
        columns=["dob"]
    )

    test = test.drop(
        columns=["dob"]
    )


print("Age feature created!")


# ============================================================
# 14. SEPARATE FEATURES AND TARGET
# ============================================================

print("\n============================================")
print("SEPARATING FEATURES AND TARGET")
print("============================================")

X_train = train.drop(
    columns=["is_fraud"]
)

y_train = train["is_fraud"]

X_test = test.drop(
    columns=["is_fraud"]
)

y_test = test["is_fraud"]


print("Training features shape:")
print(X_train.shape)

print("\nTraining target shape:")
print(y_train.shape)

print("\nTesting features shape:")
print(X_test.shape)

print("\nTesting target shape:")
print(y_test.shape)


# ============================================================
# 15. FIND CATEGORICAL COLUMNS
# ============================================================

print("\n============================================")
print("CATEGORICAL FEATURES")
print("============================================")

categorical_columns = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)


# ============================================================
# 16. ENCODE CATEGORICAL FEATURES
# ============================================================

print("\n============================================")
print("ENCODING CATEGORICAL FEATURES")
print("============================================")

for column in categorical_columns:

    # Convert values to strings

    X_train[column] = X_train[column].astype(str)

    X_test[column] = X_test[column].astype(str)


    # Create encoder

    encoder = LabelEncoder()


    # Fit only on training data

    X_train[column] = encoder.fit_transform(
        X_train[column]
    )


    # Handle unseen categories in test data

    mapping = {
        value: index
        for index, value
        in enumerate(encoder.classes_)
    }

    X_test[column] = (
        X_test[column]
        .map(mapping)
        .fillna(-1)
        .astype(int)
    )


print("Categorical features encoded successfully!")


# ============================================================
# 17. HANDLE MISSING VALUES
# ============================================================

print("\n============================================")
print("HANDLING MISSING VALUES")
print("============================================")

X_train = X_train.replace(
    [np.inf, -np.inf],
    np.nan
)

X_test = X_test.replace(
    [np.inf, -np.inf],
    np.nan
)

X_train = X_train.fillna(0)

X_test = X_test.fillna(0)

print("Missing values handled!")


# ============================================================
# 18. FEATURE SCALING
# ============================================================

print("\n============================================")
print("FEATURE SCALING")
print("============================================")

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

print("Feature scaling completed!")


# ============================================================
# 19. LOGISTIC REGRESSION
# ============================================================

print("\n============================================")
print("LOGISTIC REGRESSION")
print("============================================")

logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

print("Training Logistic Regression...")

logistic_model.fit(
    X_train_scaled,
    y_train
)

print("Logistic Regression training completed!")

print("Predicting...")

logistic_prediction = logistic_model.predict(
    X_test_scaled
)

logistic_probability = (
    logistic_model.predict_proba(
        X_test_scaled
    )[:, 1]
)

print("Logistic Regression prediction completed!")


# ============================================================
# 20. DECISION TREE
# ============================================================

print("\n============================================")
print("DECISION TREE")
print("============================================")

decision_tree = DecisionTreeClassifier(
    max_depth=15,
    class_weight="balanced",
    random_state=42
)

print("Training Decision Tree...")

decision_tree.fit(
    X_train,
    y_train
)

print("Decision Tree training completed!")

print("Predicting...")

tree_prediction = decision_tree.predict(
    X_test
)

tree_probability = (
    decision_tree.predict_proba(
        X_test
    )[:, 1]
)

print("Decision Tree prediction completed!")


# ============================================================
# 21. RANDOM FOREST
# ============================================================

print("\n============================================")
print("RANDOM FOREST")
print("============================================")

random_forest = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Training Random Forest...")

random_forest.fit(
    X_train,
    y_train
)

print("Random Forest training completed!")

print("Predicting...")

forest_prediction = random_forest.predict(
    X_test
)

forest_probability = (
    random_forest.predict_proba(
        X_test
    )[:, 1]
)

print("Random Forest prediction completed!")


# ============================================================
# 22. EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model_name,
    y_true,
    y_pred,
    y_probability
):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_true,
        y_probability
    )


    print("\n============================================")
    print(model_name)
    print("============================================")

    print(
        "Accuracy  :",
        round(accuracy, 4)
    )

    print(
        "Precision :",
        round(precision, 4)
    )

    print(
        "Recall    :",
        round(recall, 4)
    )

    print(
        "F1 Score  :",
        round(f1, 4)
    )

    print(
        "ROC-AUC   :",
        round(roc_auc, 4)
    )


    print("\nConfusion Matrix:")

    print(
        confusion_matrix(
            y_true,
            y_pred
        )
    )


    print("\nClassification Report:")

    print(
        classification_report(
            y_true,
            y_pred,
            zero_division=0
        )
    )


    return [
        model_name,
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    ]


# ============================================================
# 23. EVALUATE ALL MODELS
# ============================================================

results = []


# Logistic Regression

results.append(

    evaluate_model(

        "Logistic Regression",

        y_test,

        logistic_prediction,

        logistic_probability

    )

)


# Decision Tree

results.append(

    evaluate_model(

        "Decision Tree",

        y_test,

        tree_prediction,

        tree_probability

    )

)


# Random Forest

results.append(

    evaluate_model(

        "Random Forest",

        y_test,

        forest_prediction,

        forest_probability

    )

)


# ============================================================
# 24. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame(

    results,

    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ]

)


print("\n============================================")
print("MODEL COMPARISON")
print("============================================")

print(
    comparison.to_string(
        index=False
    )
)


# ============================================================
# 25. FIND BEST MODEL
# ============================================================

best_index = comparison[
    "F1 Score"
].idxmax()

best_model = comparison.loc[
    best_index
]


print("\n============================================")
print("BEST MODEL")
print("============================================")

print(
    "Best Model:",
    best_model["Model"]
)

print(
    "F1 Score:",
    round(best_model["F1 Score"], 4)
)

print(
    "ROC-AUC:",
    round(best_model["ROC-AUC"], 4)
)


# ============================================================
# 26. SAVE RANDOM FOREST PREDICTIONS
# ============================================================

fraud_predictions = pd.DataFrame({

    "Actual_Fraud":
        y_test.values,

    "Predicted_Fraud":
        forest_prediction,

    "Fraud_Probability":
        forest_probability

})


output_file = (
    "/content/fraud_predictions.csv"
)


fraud_predictions.to_csv(

    output_file,

    index=False

)


print("\n============================================")
print("PREDICTION FILE SAVED")
print("============================================")

print(
    output_file
)


# ============================================================
# 27. DISPLAY SAMPLE PREDICTIONS
# ============================================================

print("\n============================================")
print("SAMPLE PREDICTIONS")
print("============================================")

print(
    fraud_predictions.head(20)
)


# ============================================================
# 28. FINAL RESULT
# ============================================================

print("\n============================================")
print("TASK 2 COMPLETED")
print("============================================")

print("Credit Card Fraud Detection")

print("\nAlgorithms used:")

print("1. Logistic Regression")

print("2. Decision Tree")

print("3. Random Forest")

print("\nTarget:")

print("is_fraud")

print("\n0 = Legitimate")

print("1 = Fraudulent")

print("\nOutput file:")

print("/content/fraud_predictions.csv")

print("\nAll processing and model evaluation completed successfully!")
```
